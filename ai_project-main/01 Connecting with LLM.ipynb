### Import required libraries
This cell loads the necessary Python libraries for environment variables, API key management, and Google Gemini client.

```python
import os  # For handling environment variables
from dotenv import load_dotenv  # To load .env files
from google import genai  # Google Gemini client library
```

### Load API key from environment variables
This cell loads the Gemini API key stored in `.env` file and validates if it is available.

```python
load_dotenv(override=True)  # Load environment variables from .env file

# Get your Gemini API key from Google AI Studio
api_key = os.getenv("GEMINI_API_KEY")

# Validate if the API key is available
if api_key:
    print("Gemini API Key is available")
else:
    print("Gemini API Key is not available")
```

### Initialize Gemini Client
This cell creates a client instance for interacting with Gemini models using the API key.

```python
# Create the Gemini client using the API key
client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))
```

### Generate content from Gemini
This cell calls the Gemini model (`gemini-2.5-flash`) with a prompt and prints the generated response.

```python
# Call the model to generate content
response = client.models.generate_content(
    model="gemini-2.5-flash",  # Model name
    contents="Explain how AI works"  # Prompt to send to the model
)

# Print the response text
print(response.text)
```

### Display response with Markdown
This cell displays the model's response in formatted Markdown for better readability in Jupyter Notebook.

```python
from IPython.display import Markdown, display  # For rendering Markdown in Jupyter

# Display the response as formatted Markdown
display(Markdown(response.text))
```

### Stream response from Gemini
This cell demonstrates streaming responses from Gemini. It prints the output incrementally as chunks arrive.

```python
# Stream content from the model
response_stream = client.models.generate_content_stream(
    model="gemini-2.5-flash",  # Model name
    contents="Explain how AI works"  # Prompt
)

# Iterate over the stream and print chunks as they arrive
for chunk in response_stream:
    print(chunk.text, end="")
```
