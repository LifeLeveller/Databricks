## Let's Build the UI!

### Define a helper function
We will create a simple function `make_it_large` that takes text input, prints a log message, and returns the text in uppercase.

```python
# Define a simple function that converts input text to uppercase
def make_it_large(text):
    # Print a message to confirm the function call and input received
    print(f"Make it large has been called with input {text}")
    # Return the text in uppercase
    return text.upper()
```

### Test the helper function
Now we test the function with the input `hello`.

```python
# Call the function with sample input
make_it_large("hello")
```

## Here Comes the UI!

### Create a Gradio UI
We will use **Gradio** to build a simple web UI for the `make_it_large` function.
The input will be a text box and the output will also be a text box.

```python
import gradio as gr  # Import Gradio for UI building

# Create and launch the Gradio interface
gr.Interface(fn=make_it_large, inputs="textbox", outputs="textbox").launch()
```

### Import required libraries and environment variables
We will import `os`, `dotenv`, and `genai` to set up Gemini API usage.

```python
import os
from dotenv import load_dotenv  # For loading environment variables
from google import genai        # Google Gemini client library
```

### Load and check Gemini API key
We check if the `GEMINI_API_KEY` environment variable is available.

```python
# Load environment variables from .env file
load_dotenv(override=True)

# Get your Gemini API key from Google AI Studio
api_key = os.getenv("GEMINI_API_KEY")

# Check and confirm if the API key is available
if api_key:
    print("Gemini API Key is available")
else:
    print("Gemini API Key is not available")
```

### Create Gemini client
We initialize the Gemini client with the loaded API key.

```python
# Create the Gemini client; it automatically uses GEMINI_API_KEY from environment variables
client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))
```

### Define a function to send messages to Gemini
This function streams the response from the Gemini model and yields the result progressively.

```python
def message_gemini(input):
    # Stream content from the Gemini model
    response_stream = client.models.generate_content_stream(
        model="gemini-2.5-flash",
        contents=input
    )

    result = ""
    # Iterate over the stream and collect chunks as they arrive
    for chunk in response_stream:
        # Print each chunk to console
        print(chunk.text, end="")
        # Append chunk text to result
        result += chunk.text
        # Yield intermediate result for streaming output
        yield result
```

### Create Gradio UI for Gemini
We now build a Gradio interface for sending messages to Gemini and getting streamed responses.

```python
# Build the Gradio interface for Gemini chat
view = gr.Interface(
    fn=message_gemini,                # Function to call
    inputs=[gr.Textbox(label="Your message:")],  # Input field
    outputs=[gr.Markdown(label="Response:")],    # Output area
    flagging_mode="never"            # Disable flagging
)

# Launch the interface
view.launch()
```
